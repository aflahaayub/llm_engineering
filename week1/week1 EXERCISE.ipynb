To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
from IPython.display import Markdown, display, update_display
from openai import OpenAI
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
```

```python
# set up environment
client = OpenAI(api_key="ollama", base_url="http://localhost:11434/v1")

SYSTEM_PROMPT = """ You are a helpful AI assistant for learning AI Engineer. You understand Python, Machine Learning, Data Cleaning, Computer Vision, Speech to Speech.
You will give advice to the student with current market project example. 
"""
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
# Get gpt-4o-mini to answer, with streaming
def answer(question):
    stream = client.chat.completions.create(
        model=MODEL_LLAMA,
        messages=[
            {"role":"system", "content":SYSTEM_PROMPT},
            {"role":"user", "content":question}
        ],
        stream=True
    )
    result=""
    display_handle= display(Markdown(""), display_id=True)
    for chunk in stream:
        print(chunk.choices[0].delta)
        result+= chunk.choices[0].delta.content or ''
        update_display(Markdown(result), display_id = display_handle.display_id)
```

```python
# Get Llama 3.2 to answer
answer("if i want to switch from Java Backend Engineer to AI Engineer, what is the realistic roadmap?")
```
